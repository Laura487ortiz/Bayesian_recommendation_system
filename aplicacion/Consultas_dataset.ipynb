```python
import os
import glob
import pandas as pd

# 1. Localizar de forma automática la ruta de los archivos en Colab
files = glob.glob('/content/**/u.item', recursive=True)
if not files:
    raise FileNotFoundError("No se encontró u.item. Asegúrate de tener los archivos en la sesión.")

raw_dir = os.path.dirname(files[0])
path_data = os.path.join(raw_dir, 'u.data')
path_item = os.path.join(raw_dir, 'u.item')

# 2. Cargar tablas u.data y u.item
cols_data = ['user_id', 'movie_id', 'rating', 'timestamp']
df_ratings = pd.read_csv(path_data, sep='\t', names=cols_data)

cols_item = [
    'movie_id', 'movie_title', 'release_date', 'video_release_date', 'IMDb_URL',
    'unknown', 'Action', 'Adventure', 'Animation', 'Children', 'Comedy', 'Crime',
    'Documentary', 'Drama', 'Fantasy', 'Film-Noir', 'Horror', 'Musical', 'Mystery',
    'Romance', 'Sci-Fi', 'Thriller', 'War', 'Western'
]
df_items = pd.read_csv(path_item, sep='|', names=cols_item, encoding='latin-1')


# LOS 6 GÉNEROS MÁS POPULARES (CON MÁS PELÍCULAS)
genre_columns = [
    'Action', 'Adventure', 'Animation', 'Children', 'Comedy', 'Crime',
    'Documentary', 'Drama', 'Fantasy', 'Film-Noir', 'Horror', 'Musical',
    'Mystery', 'Romance', 'Sci-Fi', 'Thriller', 'War', 'Western'
]

# Sumar la columna binaria de cada género para contar cuántas películas pertenecen a cada uno
genre_counts = df_items[genre_columns].sum().sort_values(ascending=False)
top_6_genres = genre_counts.head(6)

print("=== TOP 6 GÉNEROS CON MÁS PELÍCULAS EN MOVIELENS ===")
print(top_6_genres)
print("\n" + "="*50 + "\n")


# CONSULTA 2: LAS 15 PELÍCULAS MÁS VISTAS / CALIFICADAS
# Contar número de evaluaciones por movie_id
top_movies = df_ratings.groupby('movie_id').agg(
    num_ratings=('rating', 'count'),
    rating_promedio=('rating', 'mean')
).reset_index()

# Unir con los títulos del catálogo u.item
top_15_movies = top_movies.merge(df_items[['movie_id', 'movie_title']], on='movie_id')
top_15_movies = top_15_movies.sort_values(by='num_ratings', ascending=False).head(15)

print("=== TOP 15 PELÍCULAS MÁS CALIFICADAS (PARA EL GOOGLE FORMS) ===")
print(top_15_movies[['movie_id', 'movie_title', 'num_ratings', 'rating_promedio']])
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid", palette="viridis")

# 1. Cargar y limpiar u.user
user_cols = ['user_id', 'age', 'gender', 'occupation', 'zip_code']
df_users = pd.read_csv('data/raw/u.user', sep='|', names=user_cols)
df_users = df_users.drop_duplicates(subset=['user_id']).copy()

# 2. Cargar y limpiar u.item
genre_cols = [
    'unknown', 'Action', 'Adventure', 'Animation', 'Children', 'Comedy',
    'Crime', 'Documentary', 'Drama', 'Fantasy', 'Film-Noir', 'Horror',
    'Musical', 'Mystery', 'Romance', 'Sci-Fi', 'Thriller', 'War', 'Western'
]
item_cols = ['movie_id', 'movie_title', 'release_date', 'video_release_date', 'imdb_url'] + genre_cols
df_items = pd.read_csv('data/raw/u.item', sep='|', encoding='latin-1', names=item_cols)

# Eliminar columnas vacías/irrelevantes, registro 'unknown' y extraer año de estreno
df_items = df_items.drop(columns=['video_release_date', 'imdb_url'])
df_items = df_items[df_items['movie_title'] != 'unknown'].copy()
df_items['release_date'] = pd.to_datetime(df_items['release_date'], errors='coerce')
df_items['release_year'] = df_items['release_date'].dt.year

# 3. Cargar y limpiar u.data
data_cols = ['user_id', 'movie_id', 'rating', 'timestamp']
df_ratings = pd.read_csv('data/raw/u.data', sep='\t', names=data_cols)
df_ratings['timestamp'] = pd.to_datetime(df_ratings['timestamp'], unit='s')

# Verificación de limpieza
print(f"Usuarios limpios: {df_users.shape} | Nulos: {df_users.isnull().sum().sum()}")
print(f"Películas limpias: {df_items.shape} | Nulos: {df_items.isnull().sum().sum()}")
print(f"Ratings limpios: {df_ratings.shape} | Nulos: {df_ratings.isnull().sum().sum()}")
```

```python
fig, axes = plt.subplots(3, 2, figsize=(16, 16))
fig.suptitle("Análisis Exploratorio — MovieLens 100K (Usuarios y Películas)", fontsize=16, fontweight='bold')

# Gráfico 1: Distribución de edad de los usuarios
sns.histplot(df_users['age'], bins=25, kde=True, ax=axes[0, 0], color='#2b5c8f')
axes[0, 0].axvline(df_users['age'].median(), color='red', linestyle='--', label=f"Mediana: {df_users['age'].median():.0f} años")
axes[0, 0].set_title("1. Distribución de Edad de Usuarios (u.user)")
axes[0, 0].set_xlabel("Edad")
axes[0, 0].set_ylabel("Cantidad de usuarios")
axes[0, 0].legend()

# Gráfico 2: Top 10 ocupaciones desagregadas por género
top_occ = df_users['occupation'].value_counts().head(10).index
sns.countplot(
    data=df_users[df_users['occupation'].isin(top_occ)],
    y='occupation', hue='gender', order=top_occ, ax=axes[0, 1]
)
axes[0, 1].set_title("2. Top 10 Ocupaciones por Género (u.user)")
axes[0, 1].set_xlabel("Cantidad de usuarios")
axes[0, 1].set_ylabel("Ocupación")

# Gráfico 3: Cantidad de películas por género en el catálogo
genre_counts = df_items[genre_cols[1:]].sum().sort_values(ascending=True)
genre_counts.plot(kind='barh', ax=axes[1, 0], color='#388e3c')
axes[1, 0].set_title("3. Cantidad de Películas por Género (u.item)")
axes[1, 0].set_xlabel("Número de películas")

# Gráfico 4: Evolución histórica de estrenos por año
movies_per_year = df_items['release_year'].value_counts().sort_index()
axes[1, 1].plot(movies_per_year.index, movies_per_year.values, color='#d95f02', linewidth=2)
axes[1, 1].fill_between(movies_per_year.index, movies_per_year.values, alpha=0.2, color='#d95f02')
axes[1, 1].set_title("4. Películas Estrenadas por Año (u.item)")
axes[1, 1].set_xlabel("Año de estreno")
axes[1, 1].set_ylabel("Cantidad de películas")

# Gráfico 5: Distribución de calificaciones (1 a 5) según género
ratings_users = df_ratings.merge(df_users[['user_id', 'gender']], on='user_id')
sns.countplot(data=ratings_users, x='rating', hue='gender', ax=axes[2, 0])
axes[2, 0].set_title("5. Distribución de Calificaciones (1-5) por Género")
axes[2, 0].set_xlabel("Calificación (Estrellas)")
axes[2, 0].set_ylabel("Total de votos")

axes[2, 1].axis('off')
plt.tight_layout(rect=[0, 0, 1, 0.97])
plt.show()
```

Limpieza Datos Encuesta

```python
# 1. Cargar archivo de encuesta con codificación latin-1 (estándar de Excel en Windows)
ruta_encuesta = "Encuesta de Preferencias Cinematográficas(1-57)(Sheet1).csv"
df_raw_survey = pd.read_csv(ruta_encuesta, sep=None, engine='python', encoding='latin-1')

# 2. Filtrar consentimiento de tratamiento de datos ("De acuerdo")
col_consentimiento = [c for c in df_raw_survey.columns if "acuerdo" in c.lower()][0]
antes = len(df_raw_survey)
df_survey = df_raw_survey[df_raw_survey[col_consentimiento].astype(str).str.strip() == "De acuerdo"].copy()
print(f"Respuestas totales: {antes} | Válidas tras filtro de consentimiento: {len(df_survey)}")

# 3. Eliminar columnas administrativas de Microsoft Forms
cols_admin = [
    'Hora de inicio', 'Hora de finalización', 'Correo electrónico',
    'Nombre', 'Total de puntos', 'Comentarios del cuestionario',
    'Hora de la última modificación', col_consentimiento
]
df_survey = df_survey.drop(columns=[c for c in cols_admin if c in df_survey.columns])

# 4. Detectar y limpiar columnas demográficas
col_edad = [c for c in df_survey.columns if "edad" in c.lower()][0]
col_genero = [c for c in df_survey.columns if "género" in c.lower() or "genero" in c.lower()][0]
col_ocupacion = [
    c for c in df_survey.columns
    if any(w in c.lower() for w in ["carrera", "programa", "ocupación", "ocupacion", "estudio", "dedica", "perfil"])
][0]

df_survey['age'] = pd.to_numeric(df_survey[col_edad], errors='coerce')
df_survey = df_survey[(df_survey['age'] >= 15) & (df_survey['age'] <= 75)].copy()
df_survey['age'] = df_survey['age'].astype(int)

def limpiar_genero(val):
    v = str(val).strip().lower()
    if v.startswith('m') or 'hombre' in v or 'masculino' in v:
        return 'M'
    elif v.startswith('f') or 'mujer' in v or 'femenino' in v:
        return 'F'
    return 'Otro'

df_survey['gender'] = df_survey[col_genero].apply(limpiar_genero)
df_survey['programa_raw'] = df_survey[col_ocupacion].astype(str).str.strip()

# Asignar user_id consecutivo a partir de 944
df_survey = df_survey.reset_index(drop=True)
df_survey['user_id'] = df_survey.index + df_users['user_id'].max() + 1

# 5. Diccionario de mapeo título -> movie_id de MovieLens 100K
mapa_peliculas = {
    'Toy Story': 1,
    'Star Wars': 50,
    'Pulp Fiction': 56,
    'Men in Black': 257,
    'Forrest Gump': 69,
    'Lion King': 71,
    'Jurassic Park': 82,
    'Home Alone': 202,
    'Aladdin': 588,
    'Scream': 288,
    'Terminator 2': 96,
    'Silence of the Lambs': 98,
    'Godfather': 127,
    'Back to the Future': 204,
    'Titanic': 313,
    'Raiders of the Lost Ark': 174,
    'Seven': 11,
    'Liar Liar': 294,
    'Mission: Impossible': 405,
    'The Rock': 117,
    'Independence Day': 121
}

rename_movies = {}
for col in df_survey.columns:
    for key, m_id in mapa_peliculas.items():
        if key.lower() in col.lower():
            rename_movies[col] = m_id
            break

print(f"Películas mapeadas correctamente: {len(rename_movies)}")
print("Categorías únicas en carrera/ocupación:", df_survey['programa_raw'].unique())

# 6. Convertir de formato ancho a largo (user_id, movie_id, rating) eliminando "NA"
df_movies_wide = df_survey[['user_id'] + list(rename_movies.keys())].rename(columns=rename_movies)
df_survey_ratings = df_movies_wide.melt(id_vars='user_id', var_name='movie_id', value_name='rating')
df_survey_ratings['movie_id'] = df_survey_ratings['movie_id'].astype(int)
df_survey_ratings['rating'] = pd.to_numeric(df_survey_ratings['rating'], errors='coerce')
df_survey_ratings = df_survey_ratings.dropna(subset=['rating']).copy()
df_survey_ratings['rating'] = df_survey_ratings['rating'].astype(int)

print(f"Total de calificaciones efectivas (sin NA): {len(df_survey_ratings)}")
```

```python
fig, axes = plt.subplots(3, 2, figsize=(16, 17))
fig.suptitle("Análisis Exploratorio — Encuesta de Preferencias Cinematográficas", fontsize=16, fontweight='bold')

# Gráfico 1: Distribución de edad de los encuestados
sns.histplot(df_survey['age'], bins=15, kde=True, ax=axes[0, 0], color='#6a1b9a')
axes[0, 0].axvline(df_survey['age'].median(), color='orange', linestyle='--', label=f"Mediana: {df_survey['age'].median():.0f} años")
axes[0, 0].set_title("1. Distribución de Edad de Encuestados")
axes[0, 0].set_xlabel("Edad")
axes[0, 0].set_ylabel("Cantidad de encuestados")
axes[0, 0].legend()

# Gráfico 2: Participación por Carrera / Ocupación y Género
sns.countplot(data=df_survey, y='programa_raw', hue='gender', ax=axes[0, 1])
axes[0, 1].set_title("2. Participación por Carrera / Ocupación y Género")
axes[0, 1].set_xlabel("Cantidad de encuestados")
axes[0, 1].set_ylabel("Carrera / Ocupación")

# Gráfico 3: Porcentaje de encuestados que vieron cada película (excluyendo NA)
tasa_vista = (df_survey_ratings.groupby('movie_id')['rating'].count() / len(df_survey) * 100).reset_index()
tasa_vista = tasa_vista.merge(df_items[['movie_id', 'movie_title']], on='movie_id').sort_values('rating', ascending=True)
axes[1, 0].barh(tasa_vista['movie_title'], tasa_vista['rating'], color='#00838f')
axes[1, 0].set_title("3. Tasa de Visualización por Película (% que la vio)")
axes[1, 0].set_xlabel("% de encuestados (sin NA)")
axes[1, 0].set_xlim(0, 105)

# Gráfico 4: Comparativa de Rating Promedio (Encuesta vs. MovieLens 100K)
avg_survey = df_survey_ratings.groupby('movie_id')['rating'].mean().rename('Encuesta')
avg_ml = df_ratings[df_ratings['movie_id'].isin(avg_survey.index)].groupby('movie_id')['rating'].mean().rename('MovieLens')
comp_df = pd.concat([avg_survey, avg_ml], axis=1).reset_index().merge(df_items[['movie_id', 'movie_title']], on='movie_id')
comp_melt = comp_df.melt(id_vars='movie_title', value_vars=['Encuesta', 'MovieLens'], var_name='Fuente', value_name='Promedio')

sns.barplot(data=comp_melt, y='movie_title', x='Promedio', hue='Fuente', ax=axes[1, 1])
axes[1, 1].set_title("4. Rating Promedio: Encuesta vs. MovieLens 100K")
axes[1, 1].set_xlabel("Calificación promedio (1-5)")
axes[1, 1].set_ylabel("")
axes[1, 1].set_xlim(0, 5.2)

# Gráfico 5: Cantidad de películas vistas por cada encuestado
vistas_por_persona = df_survey_ratings.groupby('user_id')['movie_id'].count()
sns.histplot(vistas_por_persona, bins=range(0, 22, 2), ax=axes[2, 0], color='#ef6c00')
axes[2, 0].set_title("5. Cantidad de Películas Vistas por Encuestado")
axes[2, 0].set_xlabel("Número de películas calificadas (sin NA)")
axes[2, 0].set_ylabel("Cantidad de encuestados")

axes[2, 1].axis('off')
plt.tight_layout(rect=[0, 0, 1, 0.97])
plt.show()
```
